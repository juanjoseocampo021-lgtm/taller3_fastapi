# Detección de Rostros en Python usando OpenCV

La **detección de rostros** es una de las aplicaciones más populares de la visión artificial. Consiste en identificar y localizar rostros humanos dentro de una imagen o flujo de video.

En este tutorial utilizaremos **OpenCV** y el clasificador en cascada de Haar (**Haar Cascade**), un método tradicional pero rápido y eficiente para detección en tiempo real.

---

## 🛠️ 1. Requisitos Previos

````python
# 1. instalacion de opencv
# pip install opencv-python

# 2. importacion de la libreria

import cv2

# 3.cargar imagen

imgpath = "prueba_opencv.png"

# 4. leer la imagen
img = cv2.imread(imgpath)

# 5. mostrar la imagen 
img.shape
````

````python
# 6. convertir la imagen a escala de grises
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
gray_img.shape
````

````python
# 8-
import cv2

cascade = "haarcascade_frontalface_default.xml"
face_cascade = cv2.CascadeClassifier(cascade)

if face_cascade.empty():
    print("Error: No se pudo cargar, verificar el archivo xml")
else:
    print("Clasificador cargado correctamente")
````

````python
# 9. se realiza la deteccion del rostro en la imagen en la escala de grises.
# para ello se hace un uso de la funcion detecMultiscale(), esta se utiliza para detectar
# caras  de diferentes tamaños en la imagen de entrada.

import cv2
face = face_cascade.detectMultiScale(
    gray_img, scaleFactor=1.1, minNeighbors=5, minSize=(40,40))
````

````python
#10 for para recorrer todas las caras y dibujar un rectangulo sobre la imagen original (img)


for(x, y, w, h) in face:
    cv2.rectangle(img, (x, y), (x + w, y + h), (0, 255, 0), 2)

#11. las imagenes se cargan en BGR, se debe convertir a RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

#12. importar matplotlib
import matplotlib.pyplot as plt

plt.figure(figsize=(10,10))
plt.imshow(img_rgb)
plt.axis('off')
plt.show()
````

````python
# 1. acceder a la camara de nuestro pc y leer un flujo de datos en timpo real.
video_capture = cv2.VideoCapture(0)

# 2. funcion para detectar  rostro con flujo de video y dibujar  un recuadro delimitador

def detect_bounding_box(vid):
   gray_img = cv2.cvtColor(vid, cv2.COLOR_BGR2GRAY)
   face = face_cascade.detectMultiScale(gray_img, 1.1, 5, minSize=(40,40))
   for(x, y, w, h) in face:
      cv2.rectangle(vid, (x, y), (x + w, y + h), (0, 255, 0), 2)
   return face
 
    
   
````

````python
# 3. Bucle while para detectar caras en Real time, capturar fotograma y aplicar deccion de rostros
while True:
    result, video_frame = video_capture.read()
    if not result :
        break
    faces = detect_bounding_box(video_frame)

    cv2.imshow("mi cara detectada", video_frame)
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

video_capture.release()
cv2.destroyAllWindows
````
